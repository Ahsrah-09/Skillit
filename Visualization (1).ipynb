```python
import numpy as np;
import pandas as pd;
from pandas import DataFrame,Series
import seaborn as sns;
import matplotlib.pyplot as plt
```

```python
data = sns.load_dataset("tips")
```

```python
data.head()
```

```python
#Compare the average bill of males and females
plt.figure(figsize=(6,4))
ax=sns.barplot(x=data["sex"],y=data["total_bill"])
ax.bar_label(ax.containers[0])
plt.show()
```

```python
#we want the sum of total bills of both gender female and male

agg_data = data.groupby("sex")["total_bill"].sum().reset_index()
ax=sns.barplot(x="sex",y="total_bill",data=agg_data)
ax.bar_label(ax.containers[0])
plt.show()
```

```python
ax=sns.barplot(x=data["sex"],y=data["total_bill"],hue=data["smoker"])
ax.bar_label(ax.containers[0])
ax.bar_label(ax.containers[1])
plt.show()
```

```python
# Count of smokers and non-smokers

ax = sns.countplot(x = data["smoker"])
ax.bar_label( ax.containers[0] )
```

Which region has highest and lowest orders?

```python
avg_profit = data.groupby("Region")["Profit"].mean()
avg_profit
```

```python
ax = sns.barplot( x = data["Region"]  , y = data["Profit"] )
ax.bar_label( ax.containers[0] )
plt.xticks( rotation = 45 )
plt.show()
```

```python
avg_profit = data.groupby(["Region", "Product Category"])["Profit"].mean().reset_index()
avg_profit
```

```python
# Highest average profit category in each region
result = avg_profit.loc[avg_profit.groupby("Region")["Profit"].idxmax()]
result
```

```python
plt.figure( figsize = (8,8) )
ax = sns.barplot(  x = data["Region"]  , y = data["Profit"] , hue = data["Product Category"] )
ax.bar_label( ax.containers[0] )
plt.xticks( rotation = 45 )
plt.show()
```

```python
# understand the relation between sales and profit

sns.jointplot(  x = "Profit" , y = "Sales" , data = data, kind ='scatter'  )
```
