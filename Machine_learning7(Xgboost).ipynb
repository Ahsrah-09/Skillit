```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns;
```

```python
df=pd.read_csv("/content/graduate_admission.csv")
df.head()
```

```python
df.columns
```

```python
df.drop(columns=["Unnamed: 0","Serial No."],inplace=True)
```

```python
df.head(3)
```

```python
X=df.drop(columns=["Chance of Admit "])
X.head()
```

```python
Y=df["Chance of Admit "]
Y.head()
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(X,Y,test_size=0.2,random_state=0)
```

```python
print(x_train.shape,x_test.shape,y_train.shape,y_test.shape)
```

```python
from sklearn.ensemble import RandomForestRegressor
rf=RandomForestRegressor()
rf.fit(x_train,y_train)
```

```python
y_pred=rf.predict(x_test)
```

```python
from sklearn.metrics import mean_squared_error
mse=mean_squared_error(y_test,y_pred)
print("Mean Square Error",mse)
rmse=np.sqrt(mse)
print("Root Mean Square Error",rmse)
```

```python
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
lr.fit(x_train,y_train)
```

```python
y_pred1=lr.predict(x_test)
```

```python
from sklearn.metrics import mean_squared_error
mse=mean_squared_error(y_test,y_pred1)
print("Mean Square Error of Linear Regression",mse)
rmse=np.sqrt(mse)
print("Root Mean Square Error of Linear Regression",rmse)
```

```python
from sklearn.tree import DecisionTreeRegressor
dtr=DecisionTreeRegressor()
dtr.fit(x_train,y_train)
```

```python
y_pred2=dtr.predict(x_test)
```

```python
from sklearn.metrics import mean_squared_error
mse=mean_squared_error(y_test,y_pred2)
print("Mean Square Error of Decision Tree Regression",mse)
rmse=np.sqrt(mse)
print("Root Mean Square Error of Decision Tree Regression",rmse)
```
