```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns;
```

```python
df=pd.read_csv("/content/graduate_admission.csv")
df.head()
```

```python
df.columns
```

```python
df.drop(columns=["Unnamed: 0","Serial No."],inplace=True)
```

```python
df.head(3)
```

```python
X=df.drop(columns=["Chance of Admit "])
X.head()
```

```python
Y=df["Chance of Admit "]
Y.head()
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(X,Y,test_size=0.2,random_state=0)
```

```python
print(x_train.shape,x_test.shape,y_train.shape,y_test.shape)
```

```python
from sklearn.ensemble import RandomForestRegressor
rf=RandomForestRegressor()
rf.fit(x_train,y_train)
```

```python
y_pred=rf.predict(x_test)
```

```python
from sklearn.metrics import mean_squared_error
mse=mean_squared_error(y_test,y_pred)
print("Mean Square Error",mse)
rmse=np.sqrt(mse)
print("Root Mean Square Error",rmse)
```

```python
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
lr.fit(x_train,y_train)
```

```python
y_pred1=lr.predict(x_test)
```

```python
from sklearn.metrics import mean_squared_error
mse=mean_squared_error(y_test,y_pred1)
print("Mean Square Error of Linear Regression",mse)
rmse=np.sqrt(mse)
print("Root Mean Square Error of Linear Regression",rmse)
```

```python
from sklearn.tree import DecisionTreeRegressor
dtr=DecisionTreeRegressor()
dtr.fit(x_train,y_train)
```

```python
y_pred2=dtr.predict(x_test)
```

```python
from sklearn.metrics import mean_squared_error
mse=mean_squared_error(y_test,y_pred2)
print("Mean Square Error of Decision Tree Regression",mse)
rmse=np.sqrt(mse)
print("Root Mean Square Error of Decision Tree Regression",rmse)
```

```python
%%writefile app.py
import streamlit as st
import joblib
import pandas as pd
import numpy as np

#load the trained model
model=joblib.load("/content/random.pkl")

st.title("Chance of Admission Predictor")

GRE_Score=st.number_input("Enter GRE Score",min_value=0.0,max_value=360.0,step=1.0)
Tofl_Score=st.number_input("Enter Tofl Score",min_value=0.0,max_value=120.0,step=1.0)
University_Rating=st.selectbox("Select Rating",[1,2,3,4,5])
SOP_Score=st.number_input("Enter SOP Score",min_value=0.5,max_value=5.0,step=0.5)
LOR_Score=st.number_input("Enter LOP Score",min_value=0.5,max_value=5.0,step=0.5)
CGPA=st.number_input("Enter CGPA Score",min_value=1.0,max_value=10.0,step=1.0)
Research=st.selectbox("Select a research value",[0,1])

if st.button("submit"):
  df = pd.DataFrame([[GRE_Score,Tofl_Score,University_Rating,SOP_Score,LOR_Score,CGPA,Research]])
  result = model.predict(df)[0]
  st.write("The chance of getting admitted would be ", result)
```
