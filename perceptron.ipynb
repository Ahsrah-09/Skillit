# Perceptron should make a decision weather a person gets loan or not

```python
import numpy as np
```

```python
class perceptron:
  def __init__(self,weights,bias):
    self.weights=np.array(weights)
    self.bias=bias
  def predict(self,inputs):
    z=np.dot(self.weights,np.array(inputs))+self.bias #w1*inp1+w2*inp2+b
    return 1 if z>0.5 else 0
```

```python
perp=perceptron([0.6,0.4],1)
```

```python
perp.predict([80000,8000])
```
